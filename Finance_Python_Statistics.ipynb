```python
from pandas.core.api import DataFrame
import pandas as pd
import numpy as np
import yfinance as yf
import time
!pip install matplotlib -q
from datetime import date, timedelta
from matplotlib import pyplot as plt

#asking for ticker
ticker_symbol = input("What ticker do you want? ")
print("Loading...")
time.sleep(2)
print("Data loaded!")

#loading ticker info
df = yf.download(ticker_symbol,start="2020-01-01",end="2026-11-01",auto_adjust=False,progress=False)

df['20MA'] = round(df['Close'].rolling(window=20).mean(),2)
df['20MA Yesterday'] = df['20MA'].shift(1)
df['Yesterday Close'] = df['Close'].shift(1)

#buy or sell signals in progress.
buy_signal = df['Close'] > df['20MA'] & (df['Yesterday Close']) <= df.['20MA Yesterday']
sell_signal = df['Close'] < df['20MA'] & (df['Yesterday Close']) >= df.['20MA Yesterday']

#plotting
plt.figure(figsize=(10,5))
plt.plot(df.index, df['Close'],label=str(ticker_symbol), linewidth=1, color='red')
plt.xlabel('Date',fontsize=10)
plt.ylabel('Price',fontsize=10)

plt.plot(df.index, df['20MA'],label="Moving Average 20", color='Blue')

#finalizing
plt.legend(loc='upper left', title='Legend',fontsize=9)
df.dropna()
plt.show()
df.tail(50)
```
