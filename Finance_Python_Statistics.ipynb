```python
from pandas.core.api import DataFrame
import pandas as pd
import numpy as np
import yfinance as yf
import time
!pip install matplotlib -q
from datetime import date, timedelta
from matplotlib import pyplot as plt

#asking for ticker
ticker_symbol = input("What ticker do you want? ")
print("Loading...")
time.sleep(1)
print("Data loaded!")

#loading ticker info
df = yf.download(ticker_symbol,start="2026-01-01",end="2026-11-01",auto_adjust=False,progress=False)
df.columns = df.columns.get_level_values(0)

df['200MA'] = round(df['Close'].rolling(window=200).mean(),2)

#crossover strategies
df['20MA'] = round(df['Close'].rolling(window=20).mean(),2)
df['20MA_Yesterday'] = df['20MA'].shift(1)



df['Yesterday_Close'] = df['Close'].shift(1)

#buy or sell signals in progress.
df['buy_signal'] = (df['Close'] > df['20MA']) & (df['Yesterday_Close'] <= df['20MA_Yesterday'])
df['sell_signal'] = (df['Close'] < df['20MA']) & (df['Yesterday_Close'] >= df['20MA_Yesterday'])

#plotting
plt.figure(figsize=(5,5))
plt.plot(df.index, df['Close'],label=str(ticker_symbol), linewidth=1, color='red')
plt.xlabel('Date',fontsize=10)
plt.ylabel('Price',fontsize=10)

plt.plot(df.index, df['20MA'],label="Moving Average 20", color='Blue')

buy_points = df[df['buy_signal']]
sell_points = df[df['sell_signal']]

plt.scatter(buy_points.index, buy_points['Close'], color='green',marker='^')
plt.scatter(sell_points.index, sell_points['Close'], color='red', marker='x')

#finalizing
plt.legend(loc='upper left', title='Legend',fontsize=9)
df.dropna()
plt.show()
df.tail(50)
```
