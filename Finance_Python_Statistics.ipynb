```python
from pandas.core.api import DataFrame
import pandas as pd
import numpy as np
import yfinance as yf
!pip install matplotlib -q
from datetime import date, timedelta
from matplotlib import pyplot as plt

tickers = ["META","TSLA","MSFT","NVDA"]

df = yf.download(tickers,start='2020-01-01',end='2026-10-24',progress=False)

close_prices = df['Close']
volumes = df["Volume"]

df.columns = df.columns.droplevel(1)

plt.figure(figsize=(12,5))
plt.plot(close_prices)
plt.xlabel('Year',fontsize=12)
plt.ylabel('Price',fontsize=12)
plt.legend(loc='upper left')
plt.title('Price changes',fontsize=12,fontweight='bold')

plt.figure(figsize=(12,5))
plt.scatter(close_prices.index,volumes['META'],alpha=0.5,color='Blue')
plt.xlabel('Year',fontsize=12)
plt.ylabel('Volume',fontsize=12)
plt.legend(loc='upper left')
plt.title('Volume vs year',fontsize=12,fontweight='bold')

plt.grid(False)

print(df.head())
```
